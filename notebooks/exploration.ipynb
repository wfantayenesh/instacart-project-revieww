# Instacart Data Exploration
Exploring the Instacart dataset using pandas

```python
import pandas as pd
import pathlib

# path to the csv files
raw_dir = pathlib.Path('../data/raw')

print('pandas version:', pd.__version__)
```

```python
# list of all csv files to explore
csv_files = {
    'aisles':               raw_dir / 'aisles.csv',
    'departments':          raw_dir / 'departments.csv',
    'products':             raw_dir / 'products.csv',
    'orders':               raw_dir / 'orders.csv',
    'order_products_train': raw_dir / 'order_products__train.csv',
}

# use a for loop to read all csv files
frames = {}
for name, path in csv_files.items():
    df = pd.read_csv(path)
    frames[name] = df
    print(name, '->', df.shape)

print('done loading', len(frames), 'files')
```

```python
# df.head() - show first 5 rows
for name, df in frames.items():
    print('\n---', name, '---')
    display(df.head())
```

```python
# df.shape - number of rows and columns
for name, df in frames.items():
    print(name, 'shape:', df.shape)
```

```python
# df.columns - column names
for name, df in frames.items():
    print(name, 'columns:', df.columns.tolist())
```

```python
# df.dtypes - data types
for name, df in frames.items():
    print('\n---', name, '---')
    print(df.dtypes)
```

```python
# df.describe() - statistics summary
for name, df in frames.items():
    print('\n---', name, '---')
    display(df.describe(include='all'))
```

```python
# df.info() - column info and memory usage
for name, df in frames.items():
    print('\n---', name, '---')
    df.info()
```

```python
# df.isnull().sum() - count missing values
for name, df in frames.items():
    print('\n---', name, '---')
    print(df.isnull().sum())
```

```python
# duplicate detection
for name, df in frames.items():
    dups = df.duplicated().sum()
    print(name, '- duplicate rows:', dups)
```

```python
# chunksize reading for the large prior file
# this file is 550MB so we cannot load it all at once
prior_path = raw_dir / 'order_products__prior.csv'

total = 0
chunks_read = 0

for chunk in pd.read_csv(prior_path, chunksize=500000):
    total += len(chunk)
    chunks_read += 1
    print('chunk', chunks_read, '- rows:', len(chunk), '- total so far:', total)
    if chunks_read == 1:
        print('first chunk head:')
        display(chunk.head())
        print('dtypes:', chunk.dtypes.to_dict())
        print('missing values:', chunk.isnull().sum().to_dict())

print('total rows in prior file:', total)
```
